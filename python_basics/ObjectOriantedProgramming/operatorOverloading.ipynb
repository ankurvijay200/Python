```python
class Avenger :
    def __init__(self, name, power):
        self.name = name
        self.power = power
        
    def __add__(self,other):
        return Avenger(self.name + " & " + other.name, (self.power + other.power)/2)
        
    def __sub__(self,other):
        return Avenger(self.name + " & " + other.name, (self.power - other.power)/2)
        
    def __mul__(self,other):
        return Avenger(self.name + " & " + other.name, (self.power * other.power)/2)
        
    def __truediv__(self,other):
        return Avenger(self.name + " & " + other.name, (self.power / other.power)/2)
        
    def __str__(self):
        return f"{self.name} has a power level of {self.power}"
```

```python
avengerOne = Avenger("Iron Man", 90)
avengerTwo = Avenger("Captain America", 80)
```

```python
team = avengerOne + avengerTwo
```

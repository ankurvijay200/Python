```python
## Q. convert temp from celsius to fahrenheit and vice versa using function.

def fun_temp(temp, unit):
    if unit == "C":
        result = (temp * 9/5) + 32
        print(f"{temp}°C is equal to {result}°F") 
    elif unit == "F":
        result =  (temp - 32) * 5/9
        print(f"{temp}°F is equal to {result}°C")
    else:
        return "Invalid unit. Please use 'C' for Celsius or 'F' for Fahrenheit."
```

```python
fun_temp(86, "F")
```

```python
fun_temp(30, "C")
```

```python
## Q. Write a function to check the strength of a password.

def fun2(password):
    print(f"your password is: {password}")
    percent = 0
    # Check for length
    if len(password) < 8:
        print("Password is too short. It must be at least 8 characters long.")
    else:
        percent += 20
    
    # check for uppercase letter
    if any(char.isupper() for char in password):
        percent += 20
    else:
        print("Password must contain at least one uppercase letter.")
        
    # check for lowercase letter
    if any(char.islower() for char in password):
        percent += 20
    else:
        print("Password must contain at least one lowercase letter.")
    
    # check for digit
    if any(char.isdigit() for char in password):
        percent += 20
    else:
        print("Password must contain at least one digit.")
        
    # check for special character
    if any(char in "!@#$%^&*?" for char in password):
        percent += 20
    else:
        print("Password must contain at least one special character.")
    
    print(f'your password strength is {percent}%')
```

```python
fun2(input("Enter your password: "))
```

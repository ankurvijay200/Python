```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    
    def start(self):
        pass
    
    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    
    def start(self):
        print('Car start with an ignition key')
        
    def stop(self):
        print('Car stop with break padels')
        
class ElectricScooter(Vehicle):
    
    def start(self):
        print('Scooter start with power button')
        
    def stop(self):
        print('Scooter stop with break')
```

```python
car = Car()
```

```python
car.start()
```

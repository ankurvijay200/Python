# Tuples

## can store data in ordered way just like list but tuples are not mutable

### cant change items inside tuple after once stored/inserted

```python
tuple_items = (1, 2, 3, 4, 5)
tuple_items
```

```python
tuple_items[2]
```

```python
tuple_items[2] = 10
```

## tuple also capable to store data of any data type

```python
tuple_data = ("ankur", 201, 2.5, True)
tuple_data
```

```python
tuple_data[:]
## : also can use for slicing, and all items
```

```python
tuple_data[2:]
## print all items from index 2 to end
```

```python
tuple_data[::-1]
### print all items in reverse order
```

```python
tuple_data[::-2]
### print all items in reverse order with step 2
```

# Methods in Tuple
## index()
## count()

```python
tuple_numbers = ( 1,1,2,5,7,3,4,6,9,1,3,2,6,7,8)
tuple_numbers
```

```python
tuple_numbers.index(2)
### return the index of first occurrence of 2
```

```python
tuple_numbers.count(1)
### return the count of 1 in tuple_numbers
```

# Nested Tuple

```python
tuple_nested = ((1,22,333),("Ankur","kiran"), (True,False))
print(tuple_nested[1][0])
print(not tuple_nested[2][0]) #print the opposite of first value of last tuple
```

```python
for i in tuple_nested:
    for j in i:
        print(j)
```

```python
# addition/concatenation of tuples
tuple1 = (1,2,3)
tuple2 = (14,15,16)
tuple3 = tuple1 + tuple2
print(tuple3)
```

```python
#tuple repetition
tuple1*3
#tuple1*3 is same as tuple1 + tuple1 + tuple1
```

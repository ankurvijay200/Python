# Polymorphism

# 1. Method Overriding
### Run time polymorphism

```python
class GrandParent :
    def speak(self):
        print("Hello, I'm Amit.")
        
class Parent(GrandParent) :
    def speak(self):
        print("Hello, I'm Junior Amit")
        
class Child(Parent):
    def speak(self):
        print("Hello, I'm Junior Amit Junior.")
```

```python
def attribute_provider(object):
    print(object.speak())
```

```python
granda = GrandParent()
attribute_provider(granda)
```

```python
parenta = Parent()
attribute_provider(parenta)
```

```python
childa = Child()
attribute_provider(childa)
```

# 2. Method Overloading
### Compile time polymorphism

```python
class Shape:
    def area(self):
        print('This method returns area of shape')
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return 'Area of rectangle' , self.width * self.height
```

```python
from matplotlib.pylab import float16
from numpy import double


class Circle:
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 'Area of circle', 3.14 * self.radius * self.radius
```

```python
shape = Shape()
shape.area()
```

```python
rectangle = Rectangle(2,3)
rectangle.area()
```

```python
circle = Circle(3)
circle.area()
```

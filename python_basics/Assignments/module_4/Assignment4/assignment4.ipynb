# Task 1

```python
sales = [1200, 450, 980, 1500, 3000]

f = open('sales_data.txt','w')
for sale in sales:
    f.write(str(sale) + '\n')
f.close()

f = open('sales_data.txt')
content = f.read()
print(content)
```

```python
# extra optional
sales = [1200, 450, 980, 1500, 3000]

# Open file in write mode
f = open('sales_data.txt', 'w')

# Option 1: Using string join (simplest basic approach)
f.write(','.join(str(sale) for sale in sales))

f.close()

# Read and display the output
f = open('sales_data.txt')
content = f.read()
print(content)
f.close()
```

# Task 4

```python
f = open('sales_data.txt', 'r')

lines = f.readlines()

integer_list = []

for line in lines:
    numbers = line.strip().split(',')

    for number in numbers:
        integer_list.append(int(number))

print(integer_list)

print(f"Total : {sum(integer_list)}")
print(f"Maximum : {max(integer_list)}")
print(f"Minimum : {min(integer_list)}")
print(f"Average : {sum(integer_list) / len(integer_list)}")

f.close()
```

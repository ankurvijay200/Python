## list of Squares of items 1 to 15

```python
list_squared = [ items**2 for items in range (1,16)]
list_squared_evenitems = [ items**2 for items in range (1,16) if items % 2 == 0]
```

```python
print("ankur")
print("Squares : ",list_squared)
print("Squares of even numbers : ", list_squared_evenitems)
```

## Nested List

```python
list_objects = ["Apple", "Car", "Computer", "Book", "Frame", "Spoon", "Chair"]
length_list_objects = [len(str(word)) for word in list_objects]
print(list_objects, length_list_objects)
```

```python
pair_list = [[i,j] for i in list_objects for j in length_list_objects]
pair_list
```

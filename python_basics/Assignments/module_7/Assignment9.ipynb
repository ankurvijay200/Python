```python
import numpy as np
```

```python
arr_1D = np.arange(1,11)
```

```python
arr_2D = np.arange(1,10).reshape(3, 3)
```

```python
arr_numpy = np.array([10, 20, 30, 40, 50])
```

```python
print("1D array:", arr_1D)
print("2D array:\n", arr_2D)
print("Numpy array:", arr_numpy)
```

```python
print("Shape of 1D array:", arr_1D.shape)
print("Shape of 2D array:", arr_2D.shape)
print("Shape of Numpy array:", arr_numpy.shape)
```

```python
print("1D array data type:", arr_1D.dtype)
print("2D array data type:", arr_2D.dtype)
print("Numpy array data type:", arr_numpy.dtype)
```

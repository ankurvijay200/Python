# List

## List is a built-in dynamic array which can store elements of different data types.

```python
list_varibles = ["Ankur", 1, True, 5.9]
print(list_varibles)
```

### Lists are ordered and mutable data structure

# List Indexing
## by using the proper indexing value we can retreve the data from our lists

```python
print(list_varibles[0])
```

```python
print(list_varibles)
print(list_varibles[3])
print(type(list_varibles[3]))
```

## Printing value like last , second last

```python
print(list_varibles[-1])
print(list_varibles[-2])
```

# List Slicing

```python
list_Names = ["Ankur", "Ram", "Ramesh", "Suresh", "Rakesh", "Sandeep", "Kiran"]

print(list_Names[0:3])
print(list_Names[3:-2])
```

## Reverse the list

```python
list_Names = ["Ankur", "Ram", "Ramesh", "Suresh", "Rakesh", "Sandeep", "Kiran"]

print(list_Names[:])

# reverse the list
print(list_Names[::-1]) 
```

```python
for i in list_Names:
    print(i)
```

```python
for i in list_Names:
    if(i == "Kiran"):
        print("Found Kiran")
        break;
    else:
        print("Not Found Kiran")
        
```

```python
import csv
```

```python
with open('example.txt', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Ankur', 30, 'Udaipur'])
```

```python
with open('example.txt', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## datetime Lib --> it help us for current date and time

```python
from datetime import datetime,timedelta
```

```python
datetime.now()
```

```python
# yesterday time printing
time_current = datetime.now()
yesterday = time_current - timedelta(days=1)
print(yesterday)
```

## time Lib --> it help us for current time

```python
import time
```

```python
print("Current time:", time.time())
time.sleep(5)
print("Time after 5 seconds:", time.time())
```

# Functions

### function is a block of code that is designed to perform a specific task.

### It can take inputs, called parameters, and can return an output.

### Functions help in breaking down complex problems into smaller, manageable parts, making the code more organized and reusable.

# Function syntax is as follows:
### def function_name(parameters):
###    """ function_docstring """
###    # function body
###    return [expression]

```python
def function_add(a, b):
    """This function takes two numbers as input and returns their sum."""
    return a + b

# function definition to add two numbers
```

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("Addition Of Numbers :",function_add(num1, num2));

# function call to add two numbers
```

# Recursice
### function is a function that calls itself. 

```python
# Factorial of a number using normal function
def funNum(value):
    if value == 0 or value == 1:
        return 1
    else:
        print("Fact of", value)
        count = 1
        for i in range(1, value+1):
            count *= i
        return count
```

```python
print(funNum(int(input("Enter a number to find its factorial: "))))
```

```python
def FacNum(value):
    if value == 0 or value == 1:
        return 1
    else:
       return value * FacNum(value-1)
```

```python
print("Factorial of num :",FacNum(int(input("Enter a number to find its factorial: "))))
```

# creating Simple Calculator

```python
value_1 = int(input("Enter first number: "))
value_2 = int(input("Enter second number: "))

print("Sum of two numbers is :",value_1 + value_2)
print("Subtraction of two numbers is :",value_1 - value_2)
print("Multiplication of two numbers is :",value_1 * value_2)
print("Division of two numbers is :",value_1 / value_2)
```

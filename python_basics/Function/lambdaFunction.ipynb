# Lambda Function
### Lambda functions are anonymous functions in Python that can have any number of arguments but only one expression. 

```python
# using normal function for square of a number
def square(num):
    return num * num
```

```python
square(5)
```

```python
# using lambda function for square of a number
square_lambda = lambda num: num * num
```

```python
square_lambda(5)
```

```python
type(square_lambda)
# when we check type of square_lambda, it will return <class 'function'>.
```

```python
# sum of two numbers using lambda function
sum_lambda = lambda num1, num2: num1 + num2
```

```python
sum_lambda(5, 10)
```

```python
# prodcut of 4 numbers using lambda function
product_lambda = lambda num1, num2, num3, num4: num1 * num2 * num3 * num4
```

```python
product_lambda(2, 3, 4, 5)
```

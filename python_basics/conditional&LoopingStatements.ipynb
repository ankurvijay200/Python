# Looping Statements

### looping statements can be used for iterating over a sequence (like a list, tuple, dictionary, set, or string) or other iterable objects. The two main types of loops in Python are the "for" loop and the "while" loop.

# For Loop

### for item in range(start, stop+1, step_size)

```python
list1 = []
for item in range(1,11,1):
    list1.append(item)
list1
```

```python
list1 = []
for item in range(1,21,2):
    list1.append(item)
list1
    
# for loop to print numbers from 1 to 10 and then odd numbers from 1 to 20
```

```python
name = "Ankur Vijay"
for item in name:
    print(item)
    
# for loop to print each character of a string
```

# While Loop

```python
list1 = []
i=0
while i<=10:
    list1.append(i)
    i+=1
list1   
# while loop to print numbers from 0 to 10
```

```python
list1 = []
i=0
while i<=10:
    if i%2==0:
        list1.append(i)
    else:
        pass
    i+=1
list1
# while loop to print even numbers from 0 to 10
```

```python
list1 = []
i=0
while i<=10:
    if i == 8 :
        break
    else:
        list1.append(i)
        i+=1
list1
# while loop to print numbers from 0 to 10 and break the loop when i is equal to 8
```

```python
list1 = []
i=0
while i<=20:
    if i%2==0:
        pass
    else:
        list1.append(i)
    i+=1
print(list1)
# while loop to print odd numbers from 0 to 20 and skip even numbers using pass statement
```

# Nested Loop

```python
list1 = []
for i in range(5):
    for j in range(3):
        list1.append((i, j))
list1
```

```python
list1 = []
i=0
while i <=3 : 
    j=0
    while j <= 2:
        list1.append((i, j))
        j+=1
    i+=1
list1
# nested while loop to print the value of i and j using while loop
```

```python
n = int(input("Enter a number: "))
i=0
sum = 0

while i <= n:
    sum = sum + i
    i += 1
print("The sum of first", n, "numbers is:", sum)

# while loop to calculate the sum of first n numbers entered by the user
```

```python
list1 = [1]
for i in range(2, 21):
    for j in range(2, i):
        if i % j == 0: break
    else: list1.append(i)
    
print("The prime numbers from 2 to 20 are:", list1)

# nested for loop to find prime numbers from 2 to 20 and store them in a list
```

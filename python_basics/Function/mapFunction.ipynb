# Map Function

## map function is used to apply a function to all the items in an input list (or any iterable).
### The map() function returns a map object (an iterator), which can be converted into a list or other iterables.

```python
mylist = [1, 2, 3, 4, 5]
```

```python
# using simple function Square of all numbers of list

def fun1(list1):
    empty_list = []
    for i in list1:
        empty_list.append(i**2)
    return empty_list
```

```python
print(fun1(mylist))
```

```python
# using map function with lambda function to find square of all numbers of list
print(list(map(lambda x : x*x, mylist)))
```

```python
# using map function with simple function to find square of all numbers of list
def sqre(x):
    return x*x

print(list(map(sqre, mylist)))
```

### Multiple Iterables in map function

```python
list_num1 = [1, 2, 3, 4, 5]
list_num2 = [6, 7, 8, 9, 10]

print(list(map(lambda x, y : x+y, list_num1, list_num2)))
```

```python
# using map function with simple function to convert all names of list to upper case
listName = ['ankur', 'dhoni', 'sachin', 'rohit', 'virat']
print(list(map(str.upper, listName)))
```

```python
# using map function with simple function to get all ages of employee list
list_employee = [{'name': 'ankur', 'age': 25},
                  {'name': 'dhoni', 'age': 40}, 
                  {'name': 'sachin', 'age': 50},
                  {'name': 'rohit', 'age': 30},
                  {'name': 'virat', 'age': 35}]
```

```python
def fun2(list_employee):
    return list_employee['age']
```

```python
print(list(map(fun2, list_employee)))
```

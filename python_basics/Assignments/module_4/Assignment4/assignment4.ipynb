# Task 1

```python
sales = [1200, 450, 980, 1500, 3000]

f = open('sales_data.txt','w')
for sale in sales:
    f.write(str(sale) + '\n')
f.close()

f = open('sales_data.txt')
content = f.read()
print(content)
```

```python
# extra optional
sales = [1200, 450, 980, 1500, 3000]

# Open file in write mode
f = open('sales_data.txt', 'w')

# Option 1: Using string join (simplest basic approach)
f.write(','.join(str(sale) for sale in sales))

f.close()

# Read and display the output
f = open('sales_data.txt')
content = f.read()
print(content)
f.close()
```

```python
f = open('sales_data.txt', 'r')

lines = f.readlines()

integer_list = []

for line in lines:
    numbers = line.strip().split(',')

    for number in numbers:
        integer_list.append(int(number))

print(integer_list)

print(f"Total : {sum(integer_list)}")
print(f"Maximum : {max(integer_list)}")
print(f"Minimum : {min(integer_list)}")
print(f"Average : {sum(integer_list) / len(integer_list)}")

f.close()
```

```python
import os

file_name = input("Enter the file name to read: ")

if os.path.exists(file_name):
    with open(file_name, 'r') as f:
        content = f.read()
        print(content)
else:
    print("File not found. Please check the filename.")
```

```python
# extra optional: Using try-except-finally block
file_name = input("Enter the file name to read: ")
try:
    with open(file_name, 'r') as f:
        content = f.read()
        print(content)
        f.close()
except FileNotFoundError:
    print(f"File not found. Please check the filename")
finally:
    print("Execution completed.")
```

# Task 7

```python
prices = {
    "Mouse": 500,
    "Keyboard":800,
    "Monitor": 7000,
    "pendrive":400,
    "Camera":5000
}
discount_percent = int(input("Enter the discount percentage: "))

for product, price in prices.items():
    with open('discounted_prices.txt', 'a') as f:
        discounted_price = price - (price * discount_percent / 100)
        f.write(f"Product:{product} | Orignal Price: {price} | Discounted Price: {discounted_price}\n")
        f.close()
        
with open('discounted_prices.txt', 'r') as f:
    print(f.read())
    
# optional : count total product and average discounted price
with open('discounted_prices.txt', 'r') as f:
    lines = f.readlines()
    total_products = len(lines)
    total_discounted_price = 0

    for line in lines:
        parts = line.strip().split('|')
        discounted_price = float(parts[2].split(':')[1].strip())
        total_discounted_price += discounted_price

    average_discounted_price = total_discounted_price / total_products
    print(f"Total Products: {total_products}")
    print(f"Average Discounted Price: {average_discounted_price}")
```

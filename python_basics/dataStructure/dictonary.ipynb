## Dictionary Methods
### keys()
### vaues()
### items()
### get()

```python
# keys() is used to get all the keys from the dictionary.
print("Keys : " ,dictionary.keys())

# values() is used to get all the values from the dictionary.
print("Values : " ,dictionary.values())

# items() is used to get all the items from the dictionary.
print("Items : " ,dictionary.items())

# get() is used to get the value of a specific key from the dictionary.
print("Get Name : " ,dictionary.get("Name"))
```

### shallow copy()
#### # copy() is used to create a copy of the dictionary, if we make any changes in the copy, it will not affect the original dictionary.

```python
dict1 = {"Fruit" : "Apple", "Vegetable" : "Carrot", "Grain" : "Wheat"}
```

```python
dict2 = dict1
```

```python
print("Dict1", dict1)
print("Dict2", dict2)
```

```python
dict1['Fruit'] = "Mango"
```

```python
print("Dict1", dict1)
print("Dict2", dict2)
# if we change the value of dict1, it will also change the value of dict2 because both are pointing to the same memory location.
```

```python
# so now if we want to create a copy of dict1 and assign it to dict2, we can use the copy() method so that both dict1 and dict2 are independent of each other.
dict3 = dict1.copy()
dict1['Fruit'] = "Orange"
print("Dict1", dict1)
print("Dict3", dict3)
# now if we change the value of dict1, it will not change the value of dict3 because both are independent of each other.
```

# nested dictionary 
### is use to store dictionary inside another dictionary. 


```python
dict_nested = {"Student1" : {"Name" : "Ram" , "Age" : 23, "Grade" : "A"},
          "student2" : {"Name" : "Shyam" , "Age" : 24, "Grade" : "B"},
          "student3" : {"Name" : "Mohan" , "Age" : 25, "Grade" : "C"}
          }
dict_nested
```

```python
for outer_key, outer_value in dict_nested.items():
    print("Outer Key : ", outer_key)
    for inner_key, inner_value in outer_value.items():
        print("Inner Key : ", inner_key, "Inner Value : ", inner_value)
```

## Dictionary Comprehensive

```python
comprehension_dict = {x: x**2 for x in range(11)}
comprehension_dict
# we save the square of numbers from 0 to 10 in a dictionary using dictionary comprehension.
```

```python
dict_even = {x: x**2 for x in range(1,11) if x%2 == 0}
dict_even
```

## merged Dictionary

```python
dict_a = {'a' : 1, 'b' : 2, 'c' : 3}
dict_b = {'a' : 4, 'd' : 5, 'b' : 6}
dict_c = {**dict_a, **dict_b}
dict_c
```

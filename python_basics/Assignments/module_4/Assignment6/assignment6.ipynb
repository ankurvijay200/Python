# Task 1

```python
numerator = input("Enter numerator : ")
denominator = input("Enter denominator : ")

try :
    result = float(numerator)/float(denominator)
except ValueError:
    print("Value Error")
except ZeroDivisionError :
    print("Zero division error")
else :
    print(f'Division of {numerator} and {denominator} is {result}')
finally :
    print("Operation Complete")
```

# Task 2

```python
prices = [120, 350, 'abc', 500, -200, 800]
total = 0
for price in prices:
    try:
        if not isinstance(price, (int, float)):
            raise TypeError("Value is not a number")
                
        if price < 0:
            raise ValueError("Negative price not allowed")

        total += price
        
    except (TypeError, ValueError) as e:
        print(f'Current price is {price}, {e}, Running total is {total}')
    else :  
        if(int(price) > 0 ) :
            print(f'Current price is {price}, Running total is {total}')
    
```

# Task 3

```python
def check_age(age):
    age = int(age)
    if int(age) in range(0,121):
        return "Age verified successfully."
    else :
        raise ValueError("Age must between 1 and 120")
    
try:
    print(check_age((input("Enter your age :"))))
except ValueError as e :
    print(e)
```

# Task 4

```python
try:
    filename = input("Enter the filename: ")
    with open(filename, "r") as file:
        print(f"Your file : {filename}")
        for _ in range(3):
            line = file.readline()
            if not line:
                break
            print(line, end="")
except FileNotFoundError as e:
    print(e)
except PermissionError as e:
    print(e)
finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print("File operation attempted")
```

# Task 5

```python
cart = []

while True:
    user_input = input("Enter price (or 'q' to quit): ").strip()
    if user_input.lower() == 'q':
        break
    
    try:
        price = float(user_input)
        if price < 0:
            raise Exception("Price cannot be negative.")

        cart.append(price)
    except ValueError:
        print("invalid input, please enter a number or 'q'.")
    except Exception as error:
        print(error)

print(f'cart prices : {cart}')
print(f"\ntotal items: {len(cart)}")
print(f"total bill: {sum(cart):.2f}")
```

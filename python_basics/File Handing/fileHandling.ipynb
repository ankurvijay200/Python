# File Handling

### The process of performing operations on a file, such as creating, opening, reading, writing and closing

# Need for File Handling
## Store data permanently, even after the program ends.
## Access external files like .txt, .csv, .json, etc.


# 1. Read content from file

## open() ----> Read() -----> Print() -----> close()

```python
f = open('example.txt')
content = f.read()
print(content)
f.close()
```

# 2. Write content in file

### open() -----> write() -----> close() 

#### Note : if we use 'w' mode this will replace the text
### if we use 'a' this will add text to file at end

```python
f = open('example.txt', 'w')
f.write("Hey, This is updated file")
f.close()
# this will replace the text

f = open('example.txt')
content = f.read()
print(content)
f.close()
```

```python
# add text in file , using 'a' append mode
f = open('example.txt', 'a')
f.write("\nWe used Append mode")
f.close()
# this will replace the text

f = open('example.txt')
content = f.read()
print(content)
f.close()
```

```python
lines = ['\nhello\n' , 'This\n' , 'is\n', 'Python\n', 'Course.']
with open('example.txt' , 'a') as data :
    data.writelines(lines)
    
f = open('example.txt')
content = f.read()
print(content)
f.close()
```

### write mode in binary file ----> wb
### read mode in binary ----> rb

```python
f = open('example2.txt', 'ab')
line = b'\x01\x02\x03'
f.write(line)
f.close()
```

```python
h = open('example2.txt', 'rb')
content = h.read()
print(content)
h.close()
```

### Read the content from source.txt file ----> write in new file.

```python
with open('source.txt', 'r') as source:
    content = source.read()
    
with open('destination.txt', 'w') as destination:
    destination.write(content)
    
# this autmomatically created destination file no need to create manualy
    
```

```python
# Reading total no. of line ------> total no of words ----------> total no. of characters

def lines(fileName):
    with open(fileName,'r') as files:
        lines = files.readlines()
        # readlines() function stores all line in list formate line [ 'i am ankur' , 'from ankur']
        
        total_lines = len(lines)
        # count total numaber of lines
        
        total_words = sum(len(line.split()) for line in lines)
        # count total numaber of words
        
        total_char = sum(len(line) for line in lines)
        # count total numaber of characters
        
        return total_lines, total_words, total_char
        
total_line, total_words, total_char = lines('onefile.txt')        
print(f"Total lines : {total_line} , Total Word : {total_words} , Total characters : {total_char}")
```

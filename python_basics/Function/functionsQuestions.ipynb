### Q.1 convert temp from celsius to fahrenheit and vice versa using function.

```python

def fun_temp(temp, unit):
    if unit == "C":
        result = (temp * 9/5) + 32
        print(f"{temp}°C is equal to {result}°F") 
    elif unit == "F":
        result =  (temp - 32) * 5/9
        print(f"{temp}°F is equal to {result}°C")
    else:
        return "Invalid unit. Please use 'C' for Celsius or 'F' for Fahrenheit."
```

```python
fun_temp(86, "F")
```

```python
fun_temp(30, "C")
```

## Q.2 Write a function to check the strength of a password.

```python
## Q. Write a function to check the strength of a password.

def fun2(password):
    print(f"your password is: {password}")
    percent = 0
    # Check for length
    if len(password) < 8:
        print("Password is too short. It must be at least 8 characters long.")
    else:
        percent += 20
    
    # check for uppercase letter
    if any(char.isupper() for char in password):
        percent += 20
    else:
        print("Password must contain at least one uppercase letter.")
        
    # check for lowercase letter
    if any(char.islower() for char in password):
        percent += 20
    else:
        print("Password must contain at least one lowercase letter.")
    
    # check for digit
    if any(char.isdigit() for char in password):
        percent += 20
    else:
        print("Password must contain at least one digit.")
        
    # check for special character
    if any(char in "!@#$%^&*?" for char in password):
        percent += 20
    else:
        print("Password must contain at least one special character.")
    
    print(f'your password strength is {percent}%')
```

```python
fun2(input("Enter your password: "))
```

### Q.3 check string is palindrome or not.

```python
def fun3(string):
    if string == string[::-1]:
        print(f"{string} is a palindrome.")
    else:
        print(f"{string} is not a palindrome.")
```

```python
fun3(("Mada m").lower().replace(" ", ""))
```

```python
fun3(("Ankur Vijay").lower().replace(" ", ""))
```

```python
fun3(("A man a plan a canal panama").lower().replace(" ", ""))
```

### Q.3 calculating total cost in a shopping cart

```python
cart = [
    {'item1' : 'Apple', 'price' : 10, 'quantity' : 3},
    {'item2' : 'Banana', 'price' : 5, 'quantity' : 5},
    {'item3' : 'Orange', 'price' : 8, 'quantity' : 2}
]

def fun4(cart):
    total = 0
    for item in cart:
        total += item['price'] * item['quantity']
    print(f'Total cost of items in the cart is: {total}')
    
fun4(cart)
    
```

# Getter & setter

## Getter :-

```python
class Captain:
    def __init__(self, name, address):
        self.name = name
        self.__address = address
        
    # getter method
    def accessData(self):
        print(f'My address is {self.__address}')
```

```python
capt = Captain('Captain America', 'New York')
```

```python
capt.name
```

```python
capt.address
```

```python
capt.accessData()
```

## Setter :-

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.__sound = sound
        
    def sound_info(self):
        print(f'Animal sound {self.__sound}')
        
    def sound_modify(self, new_sound):
        self.__sound = new_sound
        print(f'new sound is {self.__sound}')
```

```python
animal = Animal('Sheru','Bhao Bhao')
```

```python
animal.sound_info()
```

```python
animal.sound_modify('Meao Meao')
```

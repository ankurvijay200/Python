# Object Orianted Programming
###  approach allows developers to model real-world entities, creating programs that are more modular, reusable, and scalable compared to procedural programming. 

## A class act as a blueprint for creating objects

```python
class Cars :
    
    def __init__(self, name, brand, fuel, price, model=2026, ):
        self.name = name
        self.brand = brand
        self.fuel = fuel
        self.model = model
        self.price = price
        
    def carname(self):
        print(f"Car name is :{self.name}")
    
    def brandname(self):
        print(f"Car brand is : {self.brand}")
        
    def carfuel(self):
        print(f"Fuel type of car is : {self.fuel}")
        
    def carmodel(self):
        print(f"Car model is : {self.model}")
        
    def carprice(self):
        print(f"Car price is : {self.price}")
```

```python
bmw_x7 = Cars("X7", "BMW", "Petrol",4000000, 2025)

print(bmw_x7.carname())

print(bmw_x7.brandname())

print(bmw_x7.carfuel())

print(bmw_x7.carmodel())

print(bmw_x7.carprice())
```

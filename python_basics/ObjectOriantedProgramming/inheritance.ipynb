# 1. Single Inheritence

```python
class Avengers :
    
    def __init__(self, name, power):
        self.name = name
        self.power = power
        
    def intro(self):
        print(f'My name is {self.name}')
        
class Ironman(Avengers):
    
    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon = weapon
        
    def weapon_details(self):
        print(f'I use {self.weapon}')
    
```

```python
iron = Ironman('Ironman', 'Genius Intellect', 'Repulsor beam')
```

```python
iron.power
```

# 2. Multiple Inheritence

```python
class Car :
    
    def __init__(self, carname, transmition):
        self.carname = carname
        self.transmition = transmition
        
    def intro(self):
        print(f'Hello, This is {self.carname}')
        
    def trans(self):
        print(f'Car is {self.transmition} transmition')

class ColorDetails :
    
    def __init__(self, colorname, type):
        self.colorname = colorname
        self.type = type
        
    def colorSpacs(self):
        print(f'This car is of {self.colorname} and {self.type} finish.')
        
class FinalClass(Car, ColorDetails):
    
    def __init__(self, carname, transmition, colorname, type, seats):
        Car.__init__(self, carname, transmition)
        ColorDetails.__init__(self, colorname, type)
        self.seats = seats
        
    def seatDetails(self):
        print(f'Number of seats are {self.seats}')
```

```python
baleno = FinalClass('Venue', 'Manual', 'Black', 'Gloss', 'Seven')
```

```python
print(baleno.intro(), baleno.trans(), baleno.colorSpacs(), baleno.seatDetails())
```

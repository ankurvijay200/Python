# Types of operators

## 1. Arithmetic ---->
Addition(+), 
Substraction(-),
Product(*),
Division(/),

Floor(//) ---> remove decimal velue, 
Exponent(a**b),

## 2. Comparision
Grater than (>),
Less than (<),
Grater than or equal to (>=),
Less than or equal to (<=),
Equality Check (==),
not equal to (!=),

## 3. Logical
and -----> whether is both condition is true or not,
or ------> whether any of operation is true or not,
not -----> negation or inverse the result of given operation,



```python
# Logical Operators and
a = 10 ; b = 20 ; c = 30
a<b and b<c
```

```python
# Logical Operators or
a = 10 ; b = 20 ; c = 10
a>b or b>c
```

```python
# Logical Operators not
a = True ; b = False;
print("not of a :",not a) ; print("not of b :",not b)
```

# Exception Handing
### exception Handling lets a program detect and handle unexpected runtime errors gracefully instead of crashing abruptly, allowing it to recover or fail safely and continue execution whenever possible.

## Try-Except Blocks

### Types :
## ZeroDivisionError
## FileNotFoundError
## ValueError
## TypeError

```python
b = input("Enter the Denominator: ")
try:
    result = 1 / int(b)
    print(result)
except ZeroDivisionError as ex:
    print(ex)
except ValueError as ex:
    print(ex)
except TypeError as ex:
    print(ex)
except Exception as ex:
    print(ex)
finally:
    print("Execution completed.")
```

### file not found error

```python
try:
    f= open("requirments.txt", "r")
    content = f.read()
    print(content)
except FileNotFoundError as ex:
    print(ex)
finally:
    if 'f' in locals() and not f.closed:
        f.close()
        print("File closed successfully.")
```

# OS Library

### is able to intract with our operating system and can let us work with multiple directories and files

## getcwd() ----> is method returns the current working directory.

```python
import os
```

```python
print("The current working directory is : ", os.getcwd())
```

```python
# creating New directory
new_path = 'directoryCreateOSLearning'
os.mkdir(new_path)
print(f'New directory {new_path} created.')
```

### listdir() ----> Listing all directory present

```python
print('All items are :',os.listdir('.'))
```

### os.path.join() ----> joining multiple path together

```python
dir_name = "folder_data"
file_name = "file_data"

path = (os.getcwd(), dir_name,file_name)
print(f'The Complete path is : {path}')
```

### os.path.exists() ---> check whether a specified path exists or not.

```python
print(os.path.exists('../dataStructure/list.ipynb'))
print(os.path.exists('onefile.txt'))
print(os.path.exists('../dataStructure/ankur.txt'))
```

### complete path ----> Absloute path ------> 

```python
file_Name = 'source.txt'
complete_path = os.path.abspath(file_Name)

print(f'Complete path : {complete_path}')
```

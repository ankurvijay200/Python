# Map Function

## map function is used to apply a function to all the items in an input list (or any iterable).
### The map() function returns a map object (an iterator), which can be converted into a list or other iterables.

```python
# using simple function Square of all numbers of list

def fun1(list1):
    empty_list = []
    for i in list1:
        empty_list.append(i**2)
    return empty_list
```

```python
print(fun1([1, 2, 3, 4, 5]))
```

**NumPy** (Numerical Python) is a Python library used to perform **fast mathematical and numerical calculations**.

### **Why do we use NumPy?**

1. **Fast calculations** – Performs calculations faster than normal Python lists.
2. **Array handling** – Easily creates and manages large arrays.
3. **Mathematical operations** – Addition, subtraction, multiplication, etc.
4. **Data analysis** – Helps process large amounts of data.
5. **Scientific computing** – Used in AI, Machine Learning, and Data Science.

# Stacking

## 1. Horisontal stacking 
### Row by row joining
[ ] + [ ]

## 2. Vertical Stacking
### column by column joining

 [ ]
# +
 [ ]

```python
ary1 = np.array([[1,2],[3,4]])
ary2 = np.array([[5,6],[7,8]])
```

```python
horizontal_stack = np.hstack((ary1, ary2))
print(horizontal_stack)
```

```python
vertical_stack = np.vstack((ary1, ary2))
print(vertical_stack)
```

# Spliting
horizontal & vertical

```python
np.hsplit(horizontal_stack, 2)
```

```python
np.vsplit(vertical_stack, 4)
```

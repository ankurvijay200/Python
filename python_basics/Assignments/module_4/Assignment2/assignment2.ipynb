# Task 1

```python
def discount_fun():
    # invalid input exception handling
    try:
        order_amount = float(input("Enter order amount : "))
        if(order_amount >= 2000 ):
            discount = order_amount * 15/100
        elif(1500<=order_amount<2000):
            discount = order_amount * 10/100
        elif(1000<=order_amount<1500):
            discount = order_amount * 7/100
        else:
            discount = 0
        final_amount = order_amount - discount
        
        return order_amount, final_amount, discount
    except ValueError:
        print("Enter a valid number")
        return None
        
result = discount_fun()
# using this if becoz invalid input error occur then this will not print
if result:
    order_amount, final_amount, discount = result
    print(f'Order Amount : {order_amount} \nDiscount Amount : {discount} \nFinal amount : {final_amount}')  
```

```python
# Optional Extra (5% tax after discount)
subtotal = order_amount - discount
tax = subtotal * 5 / 100
final_total = subtotal + tax

print(f"Subtotal : {subtotal:.2f}")
print(f"Tax (5%) : {tax:.2f}")
print(f"Final Total : {final_total:.2f}")
```

```python
orders = [1200, 2500, 800, 1750, 3000]

while True:
    print("\n----- MENU -----")
    print("1. Add Order Amount")
    print("2. Show Orders and Total After Discounts")
    print("q. Quit")

    choice = input("Enter your choice: ")

    if choice == "1":
        try:
            order = int(input("Enter order amount: "))
            orders.append(order)
            print("Order added successfully!")
        except ValueError:
            print("Please enter a valid number.")
        continue

    elif choice == "2":
        if len(orders) == 0:
            print("No orders available.")
            continue

        total_revenue = 0

        print("\nOrder -> Discount% -> Final Amount")

        for item in orders:
            if item >= 2000:
                discount_percent = 15
                discount = item * 15 / 100
            elif item >= 1500:
                discount_percent = 10
                discount = item * 10 / 100
            elif item >= 1000:
                discount_percent = 7
                discount = item * 7 / 100
            else:
                discount_percent = 0
                discount = 0

            final_amount = item - discount
            total_revenue += final_amount

            print(f"{item} -> {discount_percent}% -> {final_amount:.2f}")

        print("Total Revenue:", total_revenue)
        continue

    elif choice == "q" or choice == "Q":
        print("Program Closed.")
        break

    else:
        print("Invalid choice! Please try again.")
        continue
```

```python
def discount_fun():
    # invalid input exception handling
    try:
        order_amount = int(input("Enter order amount"))
        if(order_amount >= 2000 ):
            discount = order_amount * 15/100
        elif(1500<=order_amount<2000):
            discount = order_amount * 10/100
        elif(1000<=order_amount<1500):
            discount = order_amount & 7/100
        else:
            discount = 0
        final_amount = order_amount - discount
        
        return order_amount, int(final_amount), int(discount)
    except ValueError:
        print("Enter a valid number")

result = discount_fun()
# using this if becoz invalid input error occur then this will not print
if result:
    order_amount, final_amount, discount = result
    print(f'Order Amount : {order_amount} \nDiscount Amount : {discount} \nFinal amount : {final_amount}')  
```
